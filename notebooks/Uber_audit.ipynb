```python
#standardise column names
uber_df.columns = (
    uber_df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ','_')
)
uber_df.columns
```

```python
# covert date datatype
uber_df['date'] = pd.to_datetime(uber_df['date'], format='%d/%m/%y')

#combine date + time
uber_df['booking_datetime'] = pd.to_datetime(
    uber_df['date'].astype(str) + ' ' + uber_df['time']
)

uber_df[['date','time','booking_datetime']].head()
```

```python
uber_df['booking_datetime'].dtype
```

```python
#Derived time features
uber_df['year'] = uber_df['booking_datetime'].dt.year
uber_df['month'] = uber_df['booking_datetime'].dt.month
uber_df['month_name'] = uber_df['booking_datetime'].dt.month_name()
uber_df['day'] = uber_df['booking_datetime'].dt.day
uber_df['hour'] = uber_df['booking_datetime'].dt.hour
```
